# Data Cleaning

## Bluestock Fintech Capstone Project

This notebook cleans the raw datasets by removing duplicate records, trimming text fields, and saving cleaned datasets for downstream analysis.

```python
import pandas as pd
from pathlib import Path

RAW = Path("../data/raw")
PROCESSED = Path("../data/processed")

PROCESSED.mkdir(exist_ok=True)
```

```python
files = [
    "01_fund_master.csv",
    "02_nav_history.csv",
    "03_aum_by_fund_house.csv",
    "04_monthly_sip_inflows.csv",
    "05_category_inflows.csv",
    "06_industry_folio_count.csv",
    "07_scheme_performance.csv",
    "08_investor_transactions.csv",
    "09_portfolio_holdings.csv",
    "10_benchmark_indices.csv"
]

for file in files:

    df = pd.read_csv(RAW / file)

    df.drop_duplicates(inplace=True)

    for col in df.columns:
        if df[col].dtype == "object":
            df[col] = df[col].astype(str).str.strip()

    df.to_csv(PROCESSED / f"clean_{file}", index=False)

    print(f"Cleaned {file}")
```

```python
processed_files = sorted(PROCESSED.glob("clean_*.csv"))

print(f"Total cleaned datasets: {len(processed_files)}")

for file in processed_files:
    print(file.name)
```

### Summary

- Duplicate records were removed.
- Leading and trailing spaces were trimmed from text columns.
- All cleaned datasets were saved to the `data/processed` folder for further analysis and database loading.
